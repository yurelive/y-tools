# 00 · Ambiente

Confere se o ambiente `y-tools` está pronto: versões, caminhos do projeto e onde
os modelos ficam em cache. Rode depois de `conda env create -f environment.yml`.

```python
# Versões

import platform
import sys
from importlib.metadata import PackageNotFoundError, version

print(f"Python {sys.version.split()[0]} ({platform.machine()}) em {sys.executable}")

for pacote in ["y-tools", "faster-whisper", "ctranslate2", "av", "pandas"]:
    try:
        print(f"  {pacote:<15} {version(pacote)}")
    except PackageNotFoundError:
        print(f"  {pacote:<15} ! não instalado")
```

```python
# Caminhos do projeto

from y_tools.paths import DATA_DIR, LOGS_DIR, OUTPUT_DIR, RAIZ

for nome, caminho in {"RAIZ": RAIZ, "DATA_DIR": DATA_DIR, "OUTPUT_DIR": OUTPUT_DIR, "LOGS_DIR": LOGS_DIR}.items():
    print(f"{nome:<11} {caminho}  {'✓' if caminho.exists() else '(ainda não existe)'}")
```

```python
# Hardware e modelos já baixados

import os
from pathlib import Path

import ctranslate2

print(f"Núcleos de CPU: {os.cpu_count()}")
print(f"Dispositivos CUDA: {ctranslate2.get_cuda_device_count()}  (no Mac é sempre 0 — roda na CPU)")

cache = Path.home() / ".cache" / "huggingface" / "hub"
modelos = sorted(p.name for p in cache.glob("models--*whisper*")) if cache.exists() else []
print(f"\nModelos em {cache}:")
print("\n".join(f"  {m}" for m in modelos) or "  (nenhum ainda — são baixados no primeiro uso)")
```
